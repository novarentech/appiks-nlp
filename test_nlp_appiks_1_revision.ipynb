```python
import re

def tokenize(text):
    # Lower case dulu
    text = text.lower()
    # Hapus tanda baca, ganti dengan spasi
    text = re.sub(r'[^a-z0-9\\s]', ' ', text)
    # Split jadi tokens, hapus token kosong
    tokens = [t for t in text.split() if t]
    return tokens

# Test
teks = "Aku Sedih. Tidak tahu mau apa lagi!"
tokens = tokenize(teks)
print(f"Teks asli: {teks}")
print(f"Tokens: {tokens}")
```

(NEW) Normalisasi Singkatan + Stopword Aman

```python
ABBREVIATIONS = {
    "ga": "tidak", "engga": "tidak", "kgk": "tidak", "g": "tidak",
    "gpp": "tidak apa apa", "yg": "yang", "krn": "karena",
    "jg": "juga", "tdk": "tidak", "udh": "sudah", "aja": "saja",
}

def normalize_abbreviations(text):
    for abbr, full in ABBREVIATIONS.items():
        text = re.sub(r"\b" + re.escape(abbr) + r"\b", full, text)
    return text

NEGATION_WORDS = {"tidak", "tak", "nggak", "ga", "gak", "bukan", "belum", "jangan", "enggak"}
KEYWORD_STOPWORDS = {"ada", "guna", "lagi", "saja"}
PRESERVE = NEGATION_WORDS | KEYWORD_STOPWORDS
_SASTRAWI_SW = set(StopWordRemoverFactory().get_stop_words())
STOPWORDS_SAFE = _SASTRAWI_SW - PRESERVE

print(f"Stopword Sastrawi: {len(_SASTRAWI_SW)} | dipakai (aman): {len(STOPWORDS_SAFE)}")
print("Dilindungi:", sorted(PRESERVE))
```

```python
def preprocess_appiks(text):
    text = text.lower()
    text = normalize_abbreviations(text)
    text = re.sub(r"[^a-z0-9\s]", " ", text)
    tokens = [t for t in text.split() if t]
    tokens = [t for t in tokens if t not in STOPWORDS_SAFE]
    tokens = [stemmer.stem(t) for t in tokens]
    return [t for t in tokens if t]

kalimat_test = [
    "Aku merasa sangat kesepian dan tidak berguna.",
    "Hari ini capek banget, ga ada gunanya hidup gini.",
    "Pengen mati aja rasanya, ga ada yang peduli.",
    "Aku tidak tahu harus berbuat apa lagi."
]

for k in kalimat_test:
    print(f"Input : {k}")
    print(f"Stems : {preprocess_appiks(k)}")
    print()
```

### **Weighted Scoring 3-Lapis** (revisi)


Arsitektur final Emergency Trigger = 3 lapis

Lapis 1 - Pra-pemrosesan & Pencocokan: case folding -> normalisasi singkatan -> tokenisasi -> stopword removal -> stemming Sastrawi -> pencocokan n-gram (1-3) + negation handling (skip_if_negated & require_negation).
Lapis 2 - Weighted Scoring: jumlahkan bobot seluruh matched keyword.
Lapis 3 - Hybrid Tiered Classification (Approach C): Red override / skor >= 15 -> Red / 5-14 -> Yellow / < 5 -> No Trigger.

```python
def classify_weighted(text):
    # Klasifikasi awal: weighted scoring (Lapis 2) + Approach C (Lapis 3).
    tokens = preprocess_appiks(text)

    matched = []
    total_score = 0
    has_red_keyword = False

    # Lapis 1 (sederhana, unigram) + Lapis 2 (weighted scoring)
    for token in tokens:
        if token in KAMUS:
            weight, zone, kw_type = KAMUS[token][:3]
            matched.append({"stem": token, "weight": weight, "zone": zone, "type": kw_type})
            total_score += weight
            if zone == "Red":
                has_red_keyword = True   # Approach C: ANY Red -> override

    # Lapis 3: Hybrid Tiered Classification (Approach C)
    if has_red_keyword:
        zone = "Red Zone"
        reason = "red_keyword_override"
    elif total_score >= THRESHOLD_RED_ESCALATION:
        zone = "Red Zone"
        reason = "yellow_accumulation_escalation"
    elif total_score >= THRESHOLD_YELLOW:
        zone = "Yellow Zone"
        reason = "yellow_threshold"
    else:
        zone = "No Trigger"
        reason = "below_threshold"

    breakdown = {"total_score": total_score, "has_red_keyword": has_red_keyword, "reason": reason}
    return zone, matched, total_score, breakdown
```

### **Negation Detection**

Kalimat "aku tidak mau bunuh diri" mengandung keyword Red bunuh diri tapi seharusnya tidak trigger karena di-negasi. Sebaliknya, frasa seperti "ga ada harapan" justru bermakna distress hanya ketika ada negasinya.

Logika handling per Keyword Type:

single_negative → skip jika di-negasi dalam radius 3 kata sebelumnya ("tidak mau bunuh diri" → No Trigger).


phrase_with_negation → trigger HANYA JIKA di-negasi dalam radius 3 kata sebelumnya. Stem yang disimpan adalah inti afirmatif (mis. ada harap), sehingga tanpa aturan ini "masih ada harapan" (positif) akan false positive.


single_neutral → trigger as-is.

```python
# Kata negasi Bahasa Indonesia
KATA_NEGASI = {"tidak", "tak", "nggak", "ga", "gak", "bukan", "belum", "jangan", "enggak"}
NEGATION_RADIUS = 3

def is_negated(tokens, position, radius=NEGATION_RADIUS):
    """Cek apakah token di posisi tertentu di-negasi (cek N kata sebelum)."""
    start = max(0, position - radius)
    return any(t in KATA_NEGASI for t in tokens[start:position])

# Quick test
test_tokens = ["aku", "tidak", "mau", "bunuh", "diri"]
print(f"Tokens: {test_tokens}")
print(f"'bunuh' (pos 3) negated? {is_negated(test_tokens, 3)}")  # True

test_tokens = ["aku", "mau", "bunuh", "diri", "sekarang"]
print(f"\nTokens: {test_tokens}")
print(f"'bunuh' (pos 2) negated? {is_negated(test_tokens, 2)}")  # False
```

## **Bigram Matching**

Keyword multi-kata seperti "gantung diri" tidak bisa di-match per token — harus dicek sebagai kata berurutan (bigram). Kalau tidak, kalimat "aku berdiri di sini" (stem: diri) akan false-positive.

```python
def generate_ngrams(tokens, n):
    """Generate n-gram dengan posisi awal: [(pos, "kata1 kata2"), ...]"""
    return [(i, " ".join(tokens[i:i+n])) for i in range(len(tokens) - n + 1)]

# Pisahkan kamus per panjang (1-gram, 2-gram, 3-gram)
def kamus_per_ngram(kamus):
    grouped = {1: {}, 2: {}, 3: {}}
    for stem, meta in kamus.items():
        n = len(stem.split())
        if n in grouped:
            grouped[n][stem] = meta
    return grouped

KAMUS_BY_NGRAM = kamus_per_ngram(KAMUS)
print(f"Kamus per n-gram:")
for n, d in KAMUS_BY_NGRAM.items():
    print(f"  {n}-gram: {len(d)} keyword")
```

### **Approach C — Pipeline Final**

ANY keyword Red ter-trigger → Red Zone (override)

Total skor ≥ 15 tanpa Red → Red Zone (akumulasi)

Total skor 5–14 → Yellow Zone

Total skor < 5 → No Trigger

```python
def detect_keyword(tokens, kamus_ngram, n):
    # Deteksi n-gram keyword + penanganan negasi per Keyword Type.
    #   single_negative      -> trigger KECUALI didahului negasi (skip_if_negated)
    #   phrase_with_negation -> trigger HANYA JIKA didahului negasi (require_negation)
    #   single_neutral       -> trigger apa adanya
    # Return: list dict {stem, weight, zone, category, type, position, triggered, reason}
    results = []
    for pos, ngram in generate_ngrams(tokens, n):
        if ngram in kamus_ngram:
            weight, zone, kw_type, category = kamus_ngram[ngram]
            base = {"stem": ngram, "weight": weight, "zone": zone, "category": category,
                    "type": kw_type, "position": pos}

            if kw_type == "single_negative" and is_negated(tokens, pos):
                results.append({**base, "triggered": False, "reason": "skipped_negation"})
                continue
            if kw_type == "phrase_with_negation" and not is_negated(tokens, pos):
                # Frasa inti afirmatif -> distress HANYA bila didahului negasi.
                # Tanpa aturan ini, "masih ada harapan" / "ada lagi ga?" = false positive.
                results.append({**base, "triggered": False, "reason": "requires_negation_absent"})
                continue

            results.append({**base, "triggered": True, "reason": "matched"})
    return results


def classify_final(text, verbose=False):
    #   Lapis 1 : pra-pemrosesan + pencocokan n-gram + negation handling
    #   Lapis 2 : weighted scoring
    #   Lapis 3 : Hybrid Tiered Classification (Approach C)
    tokens = preprocess_appiks(text)

    if verbose:
        print(f"Input  : {text}")
        print(f"Tokens : {tokens}")

    all_matches = []
    for n in [1, 2, 3]:
        all_matches.extend(detect_keyword(tokens, KAMUS_BY_NGRAM[n], n))

    triggered = [m for m in all_matches if m["triggered"]]
    skipped   = [m for m in all_matches if not m["triggered"]]

    total_score = sum(m["weight"] for m in triggered)
    has_red_keyword = any(m["zone"] == "Red" for m in triggered)

    if has_red_keyword:
        zone, reason = "Red Zone", "red_keyword_override"
    elif total_score >= THRESHOLD_RED_ESCALATION:
        zone, reason = "Red Zone", "yellow_accumulation_escalation"
    elif total_score >= THRESHOLD_YELLOW:
        zone, reason = "Yellow Zone", "yellow_threshold"
    else:
        zone, reason = "No Trigger", "below_threshold"

    categories = sorted({m["category"] for m in triggered})

    if verbose:
        for m in triggered:
            print(f"  [TRIGGER] {m['stem']} (bobot {m['weight']}, {m['zone']}, {m['category']})")
        for m in skipped:
            print(f"  [SKIP   ] {m['stem']} ({m['reason']})")
        print(f"  Score   : {total_score}")
        print(f"  Zone    : {zone} ({reason})")
        print(f"  Kategori: {categories}")

    return {"zone": zone, "score": total_score, "reason": reason,
            "categories": categories, "triggered": triggered, "skipped": skipped}
```

```python
import pandas as pd

labels = ["No Trigger", "Yellow Zone", "Red Zone"]
confusion = pd.DataFrame(
    [[0, 0, 0], [0, 0, 0], [0, 0, 0]],
    index=labels,
    columns=labels
)

for h in hasil_evaluasi:
    confusion.loc[h["expected"], h["predicted"]] += 1

# Tampilkan
print("CONFUSION MATRIX")
print("(baris=Expected, kolom=Predicted)")
print()
print(confusion)
```

```python
def calculate_metrics(confusion_matrix, target_class):
    tp = confusion_matrix.loc[target_class, target_class]

    # FP: prediksi target_class tapi expected bukan target_class
    fp = confusion_matrix[target_class].sum() - tp

    # FN: expected target_class tapi prediksi bukan target_class
    fn = confusion_matrix.loc[target_class].sum() - tp

    precision = tp / (tp + fp) if (tp + fp) > 0 else 0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0
    f1 = 2 * precision * recall / (precision + recall) if (precision + recall) > 0 else 0

    return precision, recall, f1, tp, fp, fn

print(f"{'Zone':<15} {'Precision':<12} {'Recall':<10} {'F1-Score':<10} {'TP':<5} {'FP':<5} {'FN':<5}")
print("-" * 75)
for zone in labels:
    p, r, f1, tp, fp, fn = calculate_metrics(confusion, zone)
    print(f"{zone:<15} {p:.2%}        {r:.2%}     {f1:.2%}     {tp:<5} {fp:<5} {fn:<5}")
```

```python
# Macro-average F1 + cek target recall Red Zone (sinkron Bab III sub-bab 3.4.2)
f1_per_zone = {}
recall_per_zone = {}
for zone in labels:
    p, r, f1, *_ = calculate_metrics(confusion, zone)
    f1_per_zone[zone] = f1
    recall_per_zone[zone] = r

macro_f1 = sum(f1_per_zone.values()) / len(labels)
red_recall = recall_per_zone["Red Zone"]
TARGET_RED_RECALL = 0.90

print(f"Macro-average F1 : {macro_f1:.2%}")
print(f"Recall Red Zone  : {red_recall:.2%}  (target >= {TARGET_RED_RECALL:.0%})")
status = "MEMENUHI" if red_recall >= TARGET_RED_RECALL else "DI BAWAH TARGET (perlu error analysis / penyesuaian kamus oleh validator)"
print(f"Status Red recall: {status}")
```

```python
from collections import Counter

# USULAN UNTUK VALIDATOR: cap kontribusi tiap keyword (atasi repetisi)
def classify_with_cap(text, max_count=None):
    # max_count=None -> tanpa batas (perilaku TA saat ini)
    # max_count=1    -> dedup (tiap keyword dihitung sekali)
    # max_count=2    -> emphasis terbatas (rekomendasi)
    tokens = preprocess_appiks(text)
    matches = []
    for n in [1, 2, 3]:
        matches.extend(detect_keyword(tokens, KAMUS_BY_NGRAM[n], n))
    triggered = [m for m in matches if m["triggered"]]

    counts = Counter(m["stem"] for m in triggered)
    weight_of = {m["stem"]: m["weight"] for m in triggered}
    zone_of = {m["stem"]: m["zone"] for m in triggered}

    total = 0
    for stem, c in counts.items():
        eff = c if max_count is None else min(c, max_count)
        total += weight_of[stem] * eff

    red = any(zone_of[s] == "Red" for s in counts)
    if red or total >= THRESHOLD_RED_ESCALATION:
        zone = "Red Zone"
    elif total >= THRESHOLD_YELLOW:
        zone = "Yellow Zone"
    else:
        zone = "No Trigger"
    return {"zone": zone, "score": total, "counts": dict(counts)}


print("Contoh: 'hampa hampa hampa hampa'")
for cap in [None, 2, 1]:
    r = classify_with_cap("hampa hampa hampa hampa", max_count=cap)
    label = "tanpa cap (TA skrg)" if cap is None else f"cap {cap}x"
    print(f"  {label:22} -> {r['zone']:12} (skor {r['score']})")

print()
print("Konstelasi keyword berbeda tetap terdeteksi (cap=2):")
for s in ["capek hidup, ga ada guna, sendiri terus, jadi beban", "mau mati aja rasanya"]:
    r = classify_with_cap(s, max_count=2)
    print(f"  {r['zone']:12} (skor {r['score']}) | {s}")
```
